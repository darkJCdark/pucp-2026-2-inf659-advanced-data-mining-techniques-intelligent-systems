Dadas las matrices $X_{test}\in\mathbb{R}^{m\times d}$ y $X_{train}\in\mathbb{R}^{n\times d}$ (con $d=784$), debes construir la matriz de distancias $D\in\mathbb{R}^{m\times n}$, donde $D_{ij}$ es la distancia al cuadrado entre la imagen de test $i$ y la imagen de entrenamiento $j$:

$$D_{ij} = \lVert x^{test}_i - x^{train}_j \rVert^2$$

**Restricción:** debes calcular $D$ de forma vectorizada, **sin bucles `for`**, usando la identidad:

$$\lVert a - b \rVert^2 = \lVert a \rVert^2 + \lVert b \rVert^2 - 2\,a\cdot b^{T}$$

Es decir, si defines los vectores de normas al cuadrado por fila de cada matriz y el producto cruzado $X_{test}\,X_{train}^{T}$, puedes obtener $D$ combinándolos con *broadcasting*.
